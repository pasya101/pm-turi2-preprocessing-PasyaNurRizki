# **C.Langkah Pratikum**

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# --- DATA MENTAH ---
data = {
    'usia': [17, 20, np.nan, 25, 30, np.nan, 35, 40, 45, 18],
    'pendapatan': [5000000, 7000000, 8000000, np.nan, 12000000, 6500000, np.nan, 4500000, 15000000, 5500000],
    'pendidikan': ['S1', 'SMA', 'S1', 'S2', 'S1', 'SMA', 'S2', 'SMA', 'S1', 'S1'],
    'kota': ['Jakarta', 'Bandung', 'Surabaya', 'Jakarta', 'Bandung', 'Jakarta', 'Surabaya', 'Bandung', 'Jakarta', 'Surabaya'],
    'membeli': ['Tidak', 'Ya', 'Ya', 'Ya', 'Ya', 'Tidak', 'Ya', 'Tidak', 'Ya', 'Tidak']
}
df = pd.DataFrame(data)

# Langkah 1 — Menangani nilai hilang
df['usia'] = df['usia'].fillna(df['usia'].median())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].median())

# Langkah 2 — Memisahkan fitur (X) dan label (y)
X = df.drop(columns=['membeli'])
y = df['membeli']

# Langkah 3 — Encoding kategorikal
X['pendidikan'] = X['pendidikan'].map({'SMA':0, 'S1':1, 'S2':2})
X = pd.get_dummies(X, columns=['kota'], dtype=int)
y = y.map({'Tidak':0, 'Ya':1})

# Langkah 4 — Membagi data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Langkah 5 — Penskalaan (StandardScaler)
num = ['usia', 'pendapatan', 'pendidikan']
sc = StandardScaler()
X_train[num] = sc.fit_transform(X_train[num])
X_test[num] = sc.transform(X_test[num])

# Langkah 6 — Verifikasi
print("--- X_train setelah preprocessing ---")
display(X_train)
```

# **D.Latihan Mandiri**

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler

# --- DATA MENTAH BARU ---
data = {
    'usia': [25, 30, np.nan, 45, 35, np.nan, 50, 23, 40, 28],
    'pendapatan': [5000000, 7000000, 8000000, np.nan, 12000000, 6500000, np.nan, 4500000, 15000000, 5500000],
    'pendidikan': ['S1', 'SMA', 'S1', 'S2', 'S1', 'SMA', 'S2', 'SMA', 'S1', 'S1'],
    'kota': ['Jakarta', 'Bandung', 'Surabaya', 'Jakarta', 'Bandung', 'Jakarta', 'Surabaya', 'Bandung', 'Jakarta', 'Surabaya'],
    # Perubahan untuk No 3: Menambahkan kolom kategorikal baru
    'jenis_kelamin': ['Pria', 'Wanita', 'Pria', 'Pria', 'Wanita', 'Pria', 'Wanita', 'Wanita', 'Pria', 'Wanita'],
    'membeli': ['Tidak', 'Ya', 'Ya', 'Ya', 'Ya', 'Tidak', 'Ya', 'Tidak', 'Ya', 'Tidak']
}
df = pd.DataFrame(data)

df['usia'] = df['usia'].fillna(df['usia'].mean())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].median()) # Pendapatan tetap median

X = df.drop(columns=['membeli'])
y = df['membeli']

X['pendidikan'] = X['pendidikan'].map({'SMA':0, 'S1':1, 'S2':2})
X = pd.get_dummies(X, columns=['kota', 'jenis_kelamin'], dtype=int)
y = y.map({'Tidak':0, 'Ya':1})

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

num = ['usia', 'pendapatan', 'pendidikan']
sc = MinMaxScaler()
X_train[num] = sc.fit_transform(X_train[num])
X_test[num] = sc.transform(X_test[num])

print("--- Tampilan 5 data teratas X_train ---")
display(X_train.head())

print("\n--- Hasil describe() X_train setelah MinMaxScaler ---")
display(X_train[num].describe())
```

Dikarenakan untuk mencegah terjadinya data leak atau kebocoran data.

2. Kapan label encoding dan kapan one-hot encoding?
-Label encoding cocok untuk data berurutan, contohnya Pendidikan (misal SD < SMP < SMA), rangking.
-Sedangkan label one-hot cocok untuk data nominal, tanpa urutan, contoh biner, nama kota.

3. Apa perbedaan normalisasi dan standardisasi?

-Nomalisai itu skala nya ditentukan dari 0-1, jika umur 17 adalah 0 dan 21 adalah 1, maka umur 18-20 akan menjadi desimal 0,(nol koma). Berguna bila butuh batas jelas

-Standarisasi itu mengubah mean 0, standar 1, tidak membatasi rentang, lebih tahan outlier -> pilihan default
